### Import necessary libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,auc,roc_auc_score,roc_curve
```

#### import dataset

```python
credit_card_data = pd.read_csv("credit_card_clean.csv")
credit_card_data
```

### data understanding

```python
credit_card_data.shape
```

```python
credit_card_data.isna().sum()
```

```python
credit_card_data.dtypes
```

### DATA PREPARATION

```python
del credit_card_data["ID"]
```

```python
credit_card_data
```

```python
le = LabelEncoder()
credit_card_data["SEX"] = le.fit_transform(credit_card_data["SEX"])
credit_card_data["EDUCATION"] = le.fit_transform(credit_card_data["EDUCATION"])
credit_card_data["MARRIAGE"] = le.fit_transform(credit_card_data["MARRIAGE"])
```

```python
credit_card_data
```

```python
credit_card_data.dtypes
```

```python
X = credit_card_data.drop("DEFAULT", axis = 1)
y = credit_card_data["DEFAULT"]
```

### Model Building

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, shuffle=True, random_state=12, stratify=y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

### model training

```python
adaboost_classifier = AdaBoostClassifier()
gradient_boost_classifier = GradientBoostingClassifier(n_estimators=50)
xgbm_classifier = XGBClassifier(n_estimators=50)
lgbm_classifier = LGBMClassifier(n_estimators=50)
```

### model testing

#### training data

```python
%%time
adaboost_classifier.fit(X_train,y_train)
```

```python
%%time
gradient_boost_classifier.fit(X_train,y_train)
```

```python
%%time
xgbm_classifier.fit(X_train,y_train)
```

```python
%%time
lgbm_classifier.fit(X_train,y_train)
```

#### AdaBoost Classifier

##### Training data

```python
y_pred_ada_train = adaboost_classifier.predict(X_train)

print("Confusion Matrix:\n", confusion_matrix(y_train,y_pred_ada_train))
print("Accuracy Score:"  ,accuracy_score(y_train,y_pred_ada_train))
print("Precision Score:" ,precision_score(y_train,y_pred_ada_train))
print("Recall Score:", recall_score(y_train,y_pred_ada_train))
```

##### Testing data

```python
y_pred_ada_test = adaboost_classifier.predict(X_test)

print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_ada_test))
print("Accuracy Score:"  ,accuracy_score(y_test,y_pred_ada_test))
print("Precision Score:" ,precision_score(y_test,y_pred_ada_test))
print("Recall Score:", recall_score(y_test,y_pred_ada_test))
```

#### Gradient Boost Classifier

```python
y_pred_gbm_train  = gradient_boost_classifier.predict(X_train)
print("Confusion Matrix:\n", confusion_matrix(y_train,y_pred_gbm_train))
print("Accuracy Score:"  ,accuracy_score(y_train,y_pred_gbm_train))
print("Precision Score:" ,precision_score(y_train,y_pred_gbm_train))
print("Recall Score:", recall_score(y_train,y_pred_gbm_train))
```

```python
y_pred_gbm_test = gradient_boost_classifier.predict(X_test)

print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_gbm_test))
print("Accuracy Score:"  ,accuracy_score(y_test,y_pred_gbm_test))
print("Precision Score:" ,precision_score(y_test,y_pred_gbm_test))
print("Recall Score:", recall_score(y_test,y_pred_gbm_test))
```

#### XGBM Classifier

##### Training data

```python
y_pred_xgbm_train = xgbm_classifier.predict(X_train)
print("Confusion Matrix:\n", confusion_matrix(y_train,y_pred_xgbm_train))
print("Accuracy Score:"  ,accuracy_score(y_train,y_pred_xgbm_train))
print("Precision Score:" ,precision_score(y_train,y_pred_xgbm_train))
print("Recall Score:", recall_score(y_train,y_pred_xgbm_train))
```
