## 1. Import Necessary Libraries

```python
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score

from datetime import datetime
```

## 2. Import Dataset

```python
breast_cancer_data = datasets.load_breast_cancer()
breast_cancer_data_df = pd.DataFrame( data=breast_cancer_data.data,columns = breast_cancer_data["feature_names"])
breast_cancer_data_df["Target"] = breast_cancer_data.target
breast_cancer_data_df.head()
```

## 3. Data Understanding

```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

## 4. Data Preparation

```python
X = breast_cancer_data_df.drop("Target",axis = 1)
```

```python
std_scaler = StandardScaler()
scaled_x = pd.DataFrame(std_scaler.fit_transform(X))
scaled_x.head()
```

```python
y = breast_cancer_data_df["Target"]
```

## 5. Model Building

```python
X_train, X_test, y_train, y_test = train_test_split(scaled_x,y,test_size=0.20,random_state=12, shuffle=True,stratify=y)
```

```python
print(X_train.shape,y_train.shape)
```

```python
print(X_test.shape, y_test.shape)
```

```python
start_time = datetime.now()
gradient_boosting_classifier = GradientBoostingClassifier(n_estimators=50,max_depth=3)
end_time = datetime.now()

print("Time taken to train the algorithm:", end_time - start_time)
```

```python
import xgboost as xgb

start_time = datetime.now()
xgb_classifier = xgb.XGBClassifier(max_depth = 3,n_estimators = 40,learning_rate = 0.008,gamma = 1.9, )
end_time = datetime.now()
print("Time Taken to Train the Algorithm:", end_time -  start_time)
```

## 6. Model Training

```python
start_time = datetime.now()
gradient_boosting_classifier.fit(X_train,y_train)
end_time = datetime.now()

print("Time taken to train the algorithm:", end_time - start_time)
```

```python
start_time = datetime.now()
xgb_classifier.fit(X_train,y_train)
end_time = datetime.now()
print("Time Taken to Train the Algorithm:", end_time -  start_time)
```

## 7. Model Testing || 8. Model Evaluation

#### Gradient Boosting Training data

```python
y_pred_train = gradient_boosting_classifier.predict(X_train)
print("Confusion Matrix:\n", confusion_matrix(y_train,y_pred_train))
print("Accuracy Score:", accuracy_score(y_train,y_pred_train))
print("Precision Score:", precision_score(y_train,y_pred_train))
print("Recall Score:", recall_score(y_train,y_pred_train))
```

#### Gradient Boosting Test data

```python
y_pred_test = gradient_boosting_classifier.predict(X_test)
print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
print("Accuracy Score:", accuracy_score(y_test,y_pred_test))
print("Precision Score:", precision_score(y_test,y_pred_test))
print("Recall Score:", recall_score(y_test,y_pred_test))
```

#### XGBM Training data

```python
y_pred_train = xgb_classifier.predict(X_train)
print("Confusion Matrix:\n", confusion_matrix(y_train,y_pred_train))
print("Accuracy Score:"  ,accuracy_score(y_train,y_pred_train))
print("Precision Score:" ,precision_score(y_train,y_pred_train))
print("Recall Score:", recall_score(y_train,y_pred_train))
```

#### XGBM Test data

```python
y_pred_test = xgb_classifier.predict(X_test)
print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
print("Accuracy Score:"  ,accuracy_score(y_test,y_pred_test))
print("Precision Score:" ,precision_score(y_test,y_pred_test))
print("Recall Score:", recall_score(y_test,y_pred_test))
```
