```python
germany_data = online_retail_data[online_retail_data["Country"] == "Germany"]
germany_data
```

```python
germany_data.shape
```

```python
germany_data.isna().sum()
```

```python
germany_pivotted_data = pd.pivot_table(data = germany_data, values="Quantity", index="InvoiceNo", columns="Description", aggfunc = 'sum').fillna(0)
germany_pivotted_data
```

```python
germany_data_onehot = germany_pivotted_data.applymap(lambda x:0 if x == 0 else 1)
germany_data_onehot
```

```python
from mlxtend.frequent_patterns import apriori,association_rules,fpgrowth
```

```python
freq_itemsets = apriori(df = germany_data_onehot,min_support=0.02,use_colnames=True,)
freq_itemsets
```

```python
best_associates = association_rules(df = freq_itemsets , metric='confidence', min_threshold=0.2)
best_associates
```

```python
best_associates.to_csv("best_anticidents_consequents.csv")
```
